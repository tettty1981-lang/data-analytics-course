При цьому функція перед діленням перевіряє, чи другий аргумент не дорівнює 0. Якщо другий аргумент не 0, фукнція вертає результат ділення, інакше - замість виконання ділення, вертає повідомлення на екран "Не можу виконати ділення на 0".

    Якщо типи, надані у функцію, не `int` - функція виводить повідомлення "Не працюю з такими типами даних.".

  2. Запустіть функцію на наступних парах чисел:
- 12 і 6
- 20 і 0
- 112 і 10
- 99 і 0
- 11 i 1.1

    і виведіть на екран повідомлення "Результат роботи функції: " - і далі те, що повернула функція.

  3. Дайте відповідь на питання: "Чому в деяких випадках функція повернула None?"

````python
def int_division(a,b):
  if not isinstance(a, int) or not isinstance(b, int):
    print("Не працюю з такими типами даних.")
    return
  if b == 0:
    print("Не можу виконати ділення на 0")
    return
  return a/b
print(int_division(12,6))
print(int_division(20,0))
print(int_division(112,10))
print(int_division(99,0))
print(int_division(11,1.1))
````

##Дайте відповідь на питання: "Чому в деяких випадках функція повернула None?"
Функція сама друкує повідомлення (print("Не можу...")) — це перший рядок

return без значення повертає None

Зовнішній print(int_division(...)) друкує те, що функція повернула — тобто None — це другий рядок
