## Перша гіпотеза: 
- може впливати ціна товару, що продається. Можемо перевірити через наші дані.
## Друга гіпотеза: 
- які саме товари(асортимент) продається. Можемо перевірити через наші дані.
## Третя гіпотеза: 
- соціально-економічні характеристики міста (чисельність населення, рівень розвитку інфраструктури й промисловості, купівельна спроможність жителів, туристична привабливість).
Цю гіпотезу можна перевірити через:
- Офіційна статистика населення міст М'янми (перепис/держстат)
- Регіональний ВВП на душу населення чи середня зарплата по регіонах
- Дані про туристичний потік (кількість відвідувачів визначних місць по містах/рік)
- Порівняти ці зовнішні показники з середнім Total/gross income по місту — чи є кореляція

Перевіряємо середню ціну по Product line відносно кожного міста

```python
df[df['City'].isin(['Mandalay', 'Yangon'])].groupby(['City', 'Product line'])['Unit price'].mean()
```

Перевіряємо частку Product line у кожному місті
